```python
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="gpt2")
```

```python
output = generator(
    "Artificial Intelligence is",
    max_new_tokens=40,
    num_return_sequences=1
)
print(output[0]['generated_text'])
```

```python
generator("Once upon a time,", max_new_tokens=40)[0]['generated_text']
```

```python
from torch.cuda import temperature
generator("In 2030, humans and AI will", max_new_tokens=50, temperature=0.9)[0]['generated_text']
```

```python
generator("Write a motivational quote", max_new_tokens=30)[0]['generated_text']
```

```python
model = pipeline("text-generation", model="google/flan-t5-small")
```

```python
model("Explain what is machine learning.")
```

```python
summarizer = pipeline("text-generation", model="facebook/bart-large-cnn")
```

```python
text = """
Artificial Intelligence is a branch of computer science that focuses on
creating systems capable of performing tasks that normally require human
intelligence. These tasks include learning, reasoning, problem solving,
understanding language, and recognizing images.
"""
summarizer(text)
```
